```python

# ============================================================
# DOWNLOAD 1-YEAR HISTORICAL DATA
# ============================================================
end_date = datetime.now()
start_date = end_date - timedelta(days=365)

tickers = df['yf'].tolist()
print(f"Period: {start_date.strftime('%Y-%m-%d')} → {end_date.strftime('%Y-%m-%d')} (1 Year)")
print(f"Downloading {len(tickers)} tickers...")

raw_data = yf.download(tickers, start=start_date, end=end_date, progress=False, auto_adjust=True)

if isinstance(raw_data.columns, pd.MultiIndex):
    price_data = raw_data['Close']
else:
    price_data = raw_data[['Close']]

# Coverage report
coverage = {}
for t in tickers:
    if t in price_data.columns:
        coverage[t] = price_data[t].dropna().__len__()
    else:
        coverage[t] = 0

valid_tickers = [t for t, d in coverage.items() if d >= 20]
missing = [t for t, d in coverage.items() if d < 20]
price_data = price_data[valid_tickers].ffill()

df['has_data'] = df['yf'].isin(valid_tickers)
print(f"✅ {len(valid_tickers)}/{len(tickers)} tickers loaded")
if missing:
    print(f"⚠️  No data: {missing}")
```

```python

# ============================================================
# BENCHMARKS + LAZY PORTFOLIOS
# ============================================================
benchmark_tickers = {'S&P 500':'SPY', 'Nasdaq 100':'QQQ', 'MSCI World':'URTH'}
bench_raw = yf.download(list(benchmark_tickers.values()), start=start_date, end=end_date,
                         progress=False, auto_adjust=True)
if isinstance(bench_raw.columns, pd.MultiIndex):
    bench_data = bench_raw['Close']
else:
    bench_data = bench_raw
bench_data.columns = list(benchmark_tickers.keys())
bench_data = bench_data.ffill().dropna()

lazy_portfolios = {
    'Ray Dalio All Weather': {'VTI':0.30,'TLT':0.40,'IEF':0.15,'GLD':0.075,'DBC':0.075},
    'Warren Buffett 90/10':  {'VOO':0.90,'BND':0.10},
    '60/40 Stocks/Bonds':    {'VTI':0.60,'BND':0.40},
    'Yale Endowment':        {'VTI':0.30,'VEA':0.15,'VWO':0.10,'VNQ':0.15,'TLT':0.15,'TIP':0.15},
    'Shiller CAPE Value':    {'VTV':0.25,'VBR':0.25,'VYM':0.25,'SCHD':0.25},
    'Cathie Wood ARK':       {'ARKK':0.40,'ARKW':0.20,'ARKF':0.20,'ARKG':0.20},
}

all_lazy_t = list(set([t for p in lazy_portfolios.values() for t in p]))
lazy_raw = yf.download(all_lazy_t, start=start_date, end=end_date, progress=False, auto_adjust=True)
if isinstance(lazy_raw.columns, pd.MultiIndex):
    lazy_prices = lazy_raw['Close']
else:
    lazy_prices = lazy_raw
lazy_prices = lazy_prices.ffill().dropna()

def calc_lp_returns(prices, weights):
    valid_w = {k:v for k,v in weights.items() if k in prices.columns}
    s = sum(valid_w.values())
    nw = {k:v/s for k,v in valid_w.items()}
    p = prices[list(nw.keys())].pct_change().dropna()
    return (p * pd.Series(nw)).sum(axis=1)

print("✅ Benchmarks and lazy portfolios loaded")
```

```python

# ============================================================
# PORTFOLIO RETURN CALCULATION
# ============================================================
valid_df = df[df['has_data']].copy()
prices = price_data[valid_df['yf'].tolist()].copy()
prices = prices.ffill().bfill().dropna(how='any')

returns = prices.pct_change().dropna()

# Weights for valid tickers (as fraction of TOTAL portfolio including cash)
weights = valid_df.set_index('yf')['weight_norm']
# Re-normalize to sum to 1 (cash treated as 0-return asset bringing rest down)
# Since we want cash to drag returns: keep normalized including cash weight
weights_for_return = weights  # already normalized vs total portfolio
# Remaining weight = cash (0 return)

portfolio_returns = (returns * weights_for_return).sum(axis=1)

# Align with benchmarks
bench_ret = bench_data.reindex(portfolio_returns.index, method='ffill').pct_change().dropna()
port_aligned = portfolio_returns.reindex(bench_ret.index).dropna()
bench_aligned = bench_ret.reindex(port_aligned.index)

# Cumulative
cum_port = (1 + port_aligned).cumprod()
cum_bench = (1 + bench_aligned).cumprod()

def calc_metrics(ret, name):
    cum = (1+ret).cumprod()
    tr = cum.iloc[-1]-1
    n = len(ret)
    cagr = (1+tr)**(252/n)-1
    vol = ret.std()*np.sqrt(252)
    sharpe = cagr/vol if vol>0 else 0
    dd = (cum/cum.cummax()-1)
    max_dd = dd.min()
    calmar = abs(cagr/max_dd) if max_dd<0 else np.nan
    down = ret[ret<0].std()*np.sqrt(252)
    sortino = cagr/down if down>0 else np.nan
    win_rate = (ret>0).mean()
    return {
        'Portfolio': name,
        'Total Return': f"{tr:.1%}",
        'CAGR': f"{cagr:.1%}",
        'Volatility': f"{vol:.1%}",
        'Sharpe': f"{sharpe:.2f}",
        'Sortino': f"{sortino:.2f}",
        'Max Drawdown': f"{max_dd:.1%}",
        'Calmar': f"{calmar:.2f}",
        'Win Rate': f"{win_rate:.0%}",
        '_tr': tr, '_cagr': cagr, '_vol': vol, '_sharpe': sharpe, '_max_dd': max_dd,
    }

m = calc_metrics(port_aligned, '⭐ YOUR PORTFOLIO')
print("=" * 55)
print("📊 PORTFOLIO PERFORMANCE — 1 YEAR")
print("=" * 55)
for k,v in m.items():
    if not k.startswith('_'):
        print(f"  {k:<20} {v}")
```
